```python
# 10. bank account
class BankAccount:
    def __init__(self, balance=0):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
        else:
            print("Insufficient balance!")
```

```python
account = BankAccount(1000)

account.deposit(500)
print(account.balance)

account.withdraw(300)
print(account.balance)
```

```python
account.withdraw(1500)
print(account.balance)
```

```python
#1. calculator
class Calculator:

    def add(self, a, b):
        return a + b

    def subtract(self, a, b):
        return a - b

    def multiply(self, a, b):
        return a * b

    def divide(self, a, b):
        if b != 0:
            return a / b
        else:
            return "Cannot divide by zero"
```

```python
calc = Calculator()
print("Addition:", calc.add(10, 5))
print("Subtraction:", calc.subtract(10, 5))
print("Multiplication:", calc.multiply(10, 5))
print("Division:", calc.divide(10, 5))
```

```python
#3 Area = pi * r * r  Circumference = 2 * pi * r
import math
class Circle:
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        return math.pi * self.radius * self.radius
    def circumference(self):
        return 2 * math.pi * self.radius
radius = float(input("Enter radius: "))
circle = Circle(radius)
print("Area:", circle.area())
print("Circumference:", circle.circumference())
```

```python
#5 Area = length × width,Perimeter = 2 × (length + width)

class Rectangle:

    def __init__(self, length, width):
        self.length = length
        self.width = width
    def area(self):
        return self.length * self.width
    def perimeter(self):
        return 2 * (self.length + self.width)
length = float(input("Enter length: "))
width = float(input("Enter width: "))
rectangle = Rectangle(length, width)
print("Area:", rectangle.area())
print("Perimeter:", rectangle.perimeter())
```

```python
#7
class Person:
    def __init__(self, name, age, address):
        self.name = name
        self.age = age
        self.address = address
    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)
        print("Address:", self.address)
name = input("Enter name: ")
age = int(input("Enter age: "))
address = input("Enter address: ")
person = Person(name, age, address)
person.display()
```

```python
#8
class Movie:
    def __init__(self, title, director, rating):
        self.title = title
        self.director = director
        self.rating = rating
    def display(self):
        print("Title:", self.title)
        print("Director:", self.director)
        print("Rating:", self.rating)
title = input("Enter movie title: ")
director = input("Enter director name: ")
rating = float(input("Enter rating: "))
movie = Movie(title, director, rating)
movie.display()
```

```python
#9
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
    def average(self):
        return sum(self.marks) / len(self.marks)
name = input("Enter student name: ")
marks = input("Enter marks: ")
marks = list(map(int, marks.split()))
student = Student(name, marks)
print( "Average Grade:", student.average())
```
