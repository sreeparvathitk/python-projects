```python
# 1.length ofthe longest word
words = ["apple", "banana", "cat", "elephant"]
longest = max(words,key=len)
print("Longest word:", longest)
print("Length:", len(longest))
```

```python
# 2.occurance of each element in a list
numbers =[1,3,2,1,2,1,1,4]
for num in set(numbers):
    print(num, ":", numbers.count(num))
```

```python
# 3.unique names
names = ["Anu", "Rahul", "Anu", "Sree", "Rahul"]
unique_names = list(set(names))
print(unique_names)
```

```python
# 4.sorting by length
words = ["apple", "cat", "elephant", "dog", "hi"]
sorted_words = sorted(words, key=len)
print(sorted_words)
```

```python
 # 5.to check if sorted
numbers = [1, 2, 3, 4, 5]
if numbers == sorted(numbers):
  print("The list is sorted")
else:
  print("The list is not sorted")
```

```python
# 6. union
list1 = [1, 2, 3, 4]
list2 = [3, 4, 5, 6]
union = list(set(list1) | set(list2))
print(union)
```

```python
# 7. dictionary
students = [
    {"name": "anu", "mark": 85},
    {"name": "appu", "mark": 92},
    {"name": "sree", "mark": 88}
]
highest = students[0]
for student in students:
    if student["mark"] > highest["mark"]:
        highest = student

print(highest)
```

```python
# 8. count in dictionary\
text = "hello"
count = {}
for char in text:
    if char in count:
        count[char] += 1
    else:
        count[char] = 1
print(count)
```

```python
#9.
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}

union = set1 | set2
intersection = set1 & set2
difference = set1 - set2

print("Union:", union)
print("Intersection:", intersection)
print("Difference:", difference)
```

```python
#11
data = [
    {"name": "sree", "marks": 80},
    {"name": "parvathi", "marks": 90},
    {"name": "Abhimanyu", "marks": 70}
]
total = sum(item["marks"] for item in data)
average = total / len(data)
print("Average:", average)
```

```python
#12.
def unique_characters(strings):
    result = set()
    for string in strings:
        result.update(string)
    return result
strings = ["hello", "world", "python"]
print(unique_characters(strings))
```
